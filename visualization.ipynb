# Figure

```python
import numpy as np
import matplotlib.pyplot as plt

# --- 설정 ---
xs = np.arange(-5, 35)
c = -0.08
lambdas = [1, 5, 15]
taus = [1, 10, 100]
lam_fixed = 15

# --- infeasible 봉우리 (barrier) ---
def penalty(x):
    return (0.1 * np.maximum(0, -(x - 0) * (x - 4)) +
            0.15 * np.maximum(0, -(x - 23/2) * (x - 18)) +
            0.05 * np.maximum(0, -(x - 22) * (x - 55/2)))

# --- feasible 영역 구불구불 지형 ---
def terrain(x):
    return (1.45 * np.maximum(0, -(x - (-4)) * (x - (-2))) +   # mode A
            0.95 * np.maximum(0, -(x - 5) * (x - 7)) +         # mode B 앞쪽
            0.95 * np.maximum(0, -(x - 8) * (x - 10)) +        # mode B 뒤쪽 (좀 높음)
            0.80 * np.maximum(0, -(x - 19) * (x - 21)) +       # mode C
            1.50 * np.maximum(0, -(x - 28) * (x - 30)) +       # mode D 앞쪽 (낮은 언덕)
            0.30 * np.maximum(0, -(x - 31) * (x - 34)))         # mode D 뒤쪽

def energy(x, lam):
    base_cost = c * x + terrain(x)
    return base_cost + lam * penalty(x)

def energy_nonneg(x, lam):
    raw = energy(x, lam)
    return raw - raw.min()

# --- 구간 정의 ---
infeasible_zones = [(0, 4), (11, 18), (22, 27)]
boundaries = [0, 4, 11, 18, 22, 27]
mode_centers = [(-1, 'A'), (9, 'B'), (20, 'C'), (31, 'D')]

def draw_zones(ax):
    ax.axvspan(xs.min(), 0, alpha=0.05, color='green')
    ax.axvspan(4, 11, alpha=0.05, color='green')
    ax.axvspan(18, 22, alpha=0.05, color='green')
    ax.axvspan(27, xs.max(), alpha=0.05, color='green')
    for a, b in infeasible_zones:
        ax.axvspan(a, b, alpha=0.06, color='red')
    for bd in boundaries:
        ax.axvline(bd, ls='--', color='gray', lw=0.8, alpha=0.5)

# def draw_mode_labels(ax):
#     ylim = ax.get_ylim()
#     y_pos = ylim[1] * 0.03 if ylim[1] > 0 else 0.5
#     for cx, label in mode_centers:
#         ax.text(cx, y_pos, f'mode {label}', ha='center',
#                 fontweight='bold', color='#534AB7', fontsize=9)

# --- 그리기 ---
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 5.5))

# ===== 왼쪽: lambda 변화 =====
colors_l = ['#3266ad', '#BA7517', '#E24B4A']
for lam, col in zip(lambdas, colors_l):
    e = energy_nonneg(xs, lam)
    ax1.plot(xs, e, 'o-', color=col, label=f'λ={lam}', markersize=5, linewidth=1.8)
    ax1.plot(xs[e.argmin()], e.min(), '*', color=col, markersize=18,
             markeredgecolor='k', markeredgewidth=0.5)

draw_zones(ax1)
ax1.set_xlabel('x (discrete)')
ax1.set_ylabel('E(x) ≥ 0')
ax1.set_title('Energy E(x) — varying λ')
ax1.legend()
ax1.set_ylim(bottom=0)
ax1.grid(alpha=0.1)
# draw_mode_labels(ax1)

# ===== 오른쪽: temperature 변화 =====
base = energy_nonneg(xs, lam_fixed)
colors_t = ['#534AB7', '#1D9E75', '#D85A30']

for tau, col in zip(taus, colors_t):
    tempered = base / tau
    ax2.plot(xs, tempered, 'o-', color=col, label=f'τ={tau}', markersize=5, linewidth=1.8)
    ax2.plot(xs[tempered.argmin()], tempered.min(), '*', color=col, markersize=18,
             markeredgecolor='k', markeredgewidth=0.5)

draw_zones(ax2)
ax2.set_xlabel('x (discrete)')
ax2.set_ylabel('E(x) / τ')
ax2.set_title(f'Tempered energy — λ={lam_fixed} fixed')
ax2.legend()
ax2.set_ylim(bottom=0)
ax2.grid(alpha=0.1)
# draw_mode_labels(ax2)

plt.tight_layout()
plt.savefig('energy_landscape.png', dpi=150, bbox_inches='tight')
plt.show()
```
