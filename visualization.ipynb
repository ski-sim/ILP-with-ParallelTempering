```python
import numpy as np
name = "sc"
node_count = 4000
heuristic = "Gurobi"
Gurobi_obj = []
time = []
for i in range(100):
    # path = f"/home/wewe1117/python_folder/CL-LNS/tmp/log/500s/{heuristic}/{name}_{node_count}/{name}_{node_count}_{i}_VANILLA_{heuristic}_nhsize0.txt"
    path = f"./discs/results_main/Gurobi/{name}_{node_count}_test/{name}_{node_count}_test_{i}_VANILLA_{heuristic}_nhsize0.txt"
    # path = f"./discs/results_main/Gurobi/{name}/{name}_{i}_VANILLA_{heuristic}_nhsize0.txt"
    obj =[]
    
    with open(path, 'r') as f:
        lines = f.readlines()
        last_line = lines[-1]
        if float(last_line.strip().split(',')[0]) <=200:
            Gurobi_obj.append(float(last_line.strip().split(',')[1]))
            
        else:
            for line in lines:
                obj.append(float(line.strip().split(',')[1]))
                if float(line.strip().split(',')[0]) >=200:
                    # print(line)
                    Gurobi_obj.append(obj[-1])
                    break

# --- 최종 결과 확인 ---
print("추출된 값의 개수:", len(Gurobi_obj))
print("추출된 값 (평균):", np.mean(Gurobi_obj)) # 처음 10개만 출력해서 확인




```
